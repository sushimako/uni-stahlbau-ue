```python
from dataclasses import dataclass
```

```python

@dataclass
class Querschnitt:
    querschnitt: str
    A: float
    Avy: float
    Avz: float
    b: float
    h: float
    tf: float
    tw: float
    Iy: float
    Iz: float
    iy: float
    iz: float
    KLy: str
    KLz: str
    Wply: float
    Wplz: float

    def um_achse(self, attr, achse):
        return self.__getattribute__(f'{attr}{achse}')

    I = lambda self, achse: self.um_achse('I', achse)
    i = lambda self, achse: self.um_achse('i', achse)
    KL = lambda self, achse: self.um_achse('KL', achse)
    Av = lambda self, achse: self.um_achse('Av', achse)
    Wpl = lambda self, achse: self.um_achse('Wpl', achse)

    @property
    def KL_BDK(self):
        if (self.h / self.b) <= 2:
            return 'b'
        else:
            return 'c'

    @property
    def hw(self):
        return self.h - 2*self.tf

    def W(self, um_achse):
        z = None
        if um_achse == 'y':
            z = self.h/2
        elif um_achse == 'z':
            z = self.b/2
        return self.I(um_achse)/z
            
```
